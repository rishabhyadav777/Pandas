```python
df["Result"]=df["Marks"].apply(
    lambda x: "Pass" if x>=45 else "Fail")

print(df)
```

```python
def student_grade(x):
    if x >= 90:
        return "A+"
    elif x >= 80:
        return "A"
    elif x >= 70:
        return "B"
    elif x >= 60:
        return "C"
    else:
        return "D"

df["Grade"] = df["Marks"].apply(student_grade)

print(df)
```
